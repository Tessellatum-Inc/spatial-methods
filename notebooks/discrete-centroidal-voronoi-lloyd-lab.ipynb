# Discrete centroidal Voronoi / Lloyd lab

Run weighted nearest-generator assignment and centroid updates on a finite point sample. This approximates a discrete CVT; it does not compute exact continuous polygon cells.

```python
import sys
sys.path.insert(0, "../implementations/python/voronoi-lloyd")
import numpy as np
from lloyd import lloyd_cvt
```

```python
points = np.array([[0.,0.],[0.,2.],[10.,0.],[10.,2.],[5.,5.]])
seeds = np.array([[1.,1.],[9.,1.]])
result = lloyd_cvt(points, seeds, max_iter=50, tolerance=1e-8)
print('generators:\n', result['centers'])
print('assignments:', result['assignments'])
print('objective:', result['objective'])
print('iterations:', result['iterations'])
print('objective history:', result['objective_history'])
```
