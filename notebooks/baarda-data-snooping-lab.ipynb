# Baarda-style data snooping lab
Fit a linear Gauss–Markov model, inspect global residual evidence and rank standardized residuals. A flagged observation is a candidate for investigation, not an automatically deleted blunder.

```python
from pathlib import Path
import sys
relative_package = Path('implementations/python/helmert-7-parameter')
repo_root = next((parent for parent in (Path.cwd(), *Path.cwd().parents)
                  if (parent / relative_package).is_dir()), None)
if repo_root is None:
    raise RuntimeError('Could not locate implementations/python/helmert-7-parameter from notebook working directory')
sys.path.insert(0, str(repo_root / relative_package))
import numpy as np
from data_snooping import adjust_and_snoop
```

```python
t=np.arange(10.0)
A=np.column_stack([np.ones_like(t),t])
y=12.0+0.8*t
y[6]+=7.0
result=adjust_and_snoop(A,y,sigma0=1.0,critical_value=3.0)
print('estimate:',result['estimate'])
print('global statistic / df:',result['global_statistic'], '/',result['global_degrees_of_freedom'])
print('standardized residuals:',np.round(result['standardized_residuals'],3))
print('suspect indices:',result['suspect_indices'])
```

## Interpretation
The global statistic measures overall disagreement with the assumed stochastic model. Local w statistics account for each observation's residual variance and network redundancy. Select a critical value for the desired significance and multiple-testing policy. Multiple blunders, correlated model errors and weak geometry can mask or smear errors; do not treat this single-outlier screen as automatic repair.
