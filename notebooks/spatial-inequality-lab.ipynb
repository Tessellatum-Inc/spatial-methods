# Spatial inequality and segregation measures

Compute area-level inequality, population entropy and two-group segregation measures. The example uses synthetic counts and values.

```python
import sys
sys.path.insert(0, "implementations/python/spatial-inequality")
from inequality import weighted_gini, theil_t, theil_l, spatial_entropy, dissimilarity_index, exposure_indices
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Weighted inequality across area values

Gini, Theil T and Theil L respond differently to the distribution's tails. Population weights are used here.

```python
values = [12, 18, 24, 40, 55]
population = [100, 200, 150, 50, 25]
print('weighted Gini:', weighted_gini(values, population))
print('Theil T:', theil_t(values, population))
print('Theil L:', theil_l(values, population))
```

## 2. Entropy of population across areas

This is the Shannon entropy of area population shares. It measures evenness across the supplied units, not spatial clustering or adjacency.

```python
entropy = spatial_entropy(population)
print(entropy)
plt.bar(np.arange(len(population)), population)
plt.xlabel('Area index'); plt.ylabel('Population'); plt.title('Population across areas'); plt.show()
```

## 3. Two-group segregation

Dissimilarity measures evenness; isolation and interaction measure potential exposure.

```python
group_a = [80, 20, 30, 70]
group_b = [20, 80, 70, 30]
print('dissimilarity:', dissimilarity_index(group_a, group_b))
print('exposure:', exposure_indices(group_a, group_b))
```
