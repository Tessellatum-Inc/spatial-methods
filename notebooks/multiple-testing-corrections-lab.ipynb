# Multiple-testing corrections

Compare family-wise error-rate and false-discovery-rate procedures on one synthetic p-value family.

```python
import sys
sys.path.insert(0, "implementations/python/multiple-testing")
from multiple_testing import adjust_pvalues
import pandas as pd
```

## 1. Define the hypothesis family

The supplied p-values form one family. The family definition is not inferred from adjacency or variable names.

```python
pvalues = [0.001, 0.006, 0.02, 0.03, 0.08, 0.12, 0.2]
methods = ['bonferroni', 'sidak', 'holm', 'hochberg', 'bh', 'by']
rows = []
for method in methods:
    result = adjust_pvalues(pvalues, method=method, alpha=0.05)
    rows.append({'method': method, 'adjusted_pvalues': result['adjusted_pvalues'], 'rejections': sum(result['reject'])})
pd.DataFrame(rows)
```

## 2. Compare the error criteria

FWER procedures target the probability of at least one false rejection. FDR procedures target the expected false-discovery proportion. BH's dependence assumptions matter for spatial tests; BY is more conservative but valid under arbitrary dependence.
