# Bivariate Moran's I lab
Compare a variable at each location with the spatial lag of a second variable. The statistic is directional: swapping the variables can change the result.

```python
import sys
sys.path.insert(0, '../implementations/python/bivariate-morans-i')
from bivariate_morans import bivariate_morans_i
```

## 1. Define a small spatial system
Four units form a chain. The weights are binary in the input; the implementation row-standardizes each non-island row.

```python
weights = [
    [0, 1, 0, 0],
    [1, 0, 1, 0],
    [0, 1, 0, 1],
    [0, 0, 1, 0],
]
x = [1, 2, 3, 4]  # variable measured at each unit
y = [2, 1, 4, 3]  # second variable, measured at the same units
```

## 2. Calculate the directional association
The implementation standardizes both variables with the sample standard deviation and calculates x' W y / (n - 1).

```python
xy = bivariate_morans_i(x, y, weights, permutations=499, seed=2026)
yx = bivariate_morans_i(y, x, weights, permutations=499, seed=2026)
print('I_xy:', round(xy['I'], 4), 'conditional pseudo-p:', round(xy['p_sim'], 4))
print('I_yx:', round(yx['I'], 4), 'conditional pseudo-p:', round(yx['p_sim'], 4))
```

## 3. Interpret the result
The order of the variables matters. The first statistic relates x at each unit to neighbouring y values; the reverse statistic relates y to neighbouring x values. Neither result is causal. The permutation scheme holds x and W fixed while shuffling y, so report that conditional null and the chosen weights transformation.

```python
# Inspect the simulated reference distribution for the first direction.
print('Permutation mean:', round(xy['mean_sim'], 4))
print('Permutation SD:', round(xy['sd_sim'], 4))
print('Number of permutations:', xy['permutations'])
```
