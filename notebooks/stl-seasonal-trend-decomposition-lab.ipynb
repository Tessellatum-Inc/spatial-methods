# STL-style seasonal–trend decomposition lab

This notebook explores the repository's compact educational implementation. It is not a full reproduction of Cleveland et al. (1990).

```python
import sys
sys.path.insert(0, "implementations/python/temporal-models")
import math
import matplotlib.pyplot as plt
from stl import stl_decompose

period = 12
y = [0.04*t + 2*math.sin(2*math.pi*t/period) + 0.5*math.cos(4*math.pi*t/period) for t in range(120)]
y[61] += 15  # isolated anomaly
```

```python
fit = stl_decompose(y, period, seasonal_window=7, trend_window=21, robust=True)
fig, axes = plt.subplots(4, 1, figsize=(11, 8), sharex=True)
for ax, (name, values) in zip(axes, [('Observed', y), ('Trend', fit['trend']), ('Seasonal', fit['seasonal']), ('Remainder', fit['remainder'])]):
    ax.plot(values); ax.set_title(name)
plt.tight_layout()
```
