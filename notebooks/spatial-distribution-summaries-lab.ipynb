# Geographic distribution summaries

Compare weighted mean center, geometric median center, standard distance and a covariance-based standard deviational ellipse. Coordinates are projected planar coordinates.

```python
import sys
sys.path.insert(0, "implementations/python/spatial-distribution-summaries")
from spatial_centers import mean_center, median_center, standard_distance, standard_deviational_ellipse
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Ellipse
```

## 1. Mean and median centers

The median center minimizes total Euclidean distance and is less sensitive to peripheral outliers.

```python
points = [(0,0),(1,1),(2,0),(1,2),(10,8)]
weights = [1,2,1,2,0.5]
mean = mean_center(points, weights)
median = median_center(points, weights)
print('weighted mean center:', mean)
print('weighted median center:', median)
```

## 2. Standard distance and ellipse

The ellipse uses covariance eigenvectors and eigenvalues. Its orientation is counterclockwise from the positive x-axis. A scale of approximately 1.515 produces the joint 68-percent ellipse under a bivariate normal model.

```python
sd = standard_distance(points, weights)
ellipse = standard_deviational_ellipse(points, weights, scale=1.515)
print('standard distance:', sd)
print('ellipse summary:', ellipse)
fig, ax = plt.subplots(figsize=(7, 6))
xy = np.asarray(points)
ax.scatter(xy[:,0], xy[:,1], s=np.asarray(weights)*40, label='weighted locations')
ax.scatter(*mean, marker='x', s=100, label='mean center')
ax.scatter(*median, marker='+', s=120, label='median center')
patch = Ellipse(ellipse['center'], 2*ellipse['major_axis_sd'], 2*ellipse['minor_axis_sd'], angle=ellipse['orientation_degrees_ccw_from_x'], fill=False, linewidth=2, label='68% Gaussian ellipse')
ax.add_patch(patch)
ax.set_aspect('equal'); ax.legend(); ax.set_title('Centers and directional spread'); plt.show()
```
