# Mantel and partial Mantel lab
Compare aligned distance matrices, then inspect the limits of the permutation null.

```python
import sys
sys.path.insert(0, '../implementations/python/mantel')
from mantel import mantel_test, partial_mantel_test
```

## 1. Three aligned distance matrices
Every matrix uses the same observation order.

```python
dx=[[0,1,2,3],[1,0,1,2],[2,1,0,1],[3,2,1,0]]
dy=[[0,2,4,6],[2,0,2,4],[4,2,0,2],[6,4,2,0]]
dz=[[0,3,1,4],[3,0,2,1],[1,2,0,3],[4,1,3,0]]
print(mantel_test(dx,dy,permutations=499,seed=42))
print(partial_mantel_test(dx,dy,dz,permutations=499,seed=42))
```

## 2. What is permuted?
A single permutation relabels observations, so the same permutation is applied to rows and columns. The matrix is not flattened and shuffled cell by cell.

```python
print('simple:',mantel_test(dx,dy,permutations=999,seed=7))
print('partial:',partial_mantel_test(dx,dy,dz,permutations=999,seed=7))
```

## 3. Interpretation checkpoint
The partial statistic removes a linear relationship with the control distance vector. It does not generally control spatial autocorrelation. Conventional p-values can be anti-conservative when both tested variables are spatially structured.
