# Point-pattern KDE: how bandwidth changes the intensity surface

This lab turns a synthetic point pattern into a continuous Gaussian kernel intensity surface. It is descriptive, not a statistical test of clustering.

```python
from pathlib import Path
import sys
import numpy as np
import matplotlib.pyplot as plt
ROOT = Path.cwd()
if not (ROOT / 'implementations').exists(): ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'implementations/python/point-pattern-kde'))
from point_pattern_kde import gaussian_kde, adaptive_gaussian_kde, select_bandwidth_loo, silverman_bandwidth
```

## 1. Create a reproducible point pattern

Two synthetic event concentrations are placed inside a rectangular study window. The coordinates are planar teaching coordinates, not a real geographic dataset.

```python
rng = np.random.default_rng(17)
left = rng.normal(loc=(3, 3), scale=(0.65, 0.65), size=(45, 2))
right = rng.normal(loc=(7, 7), scale=(0.8, 0.8), size=(45, 2))
points = np.vstack([left, right])
points = points[((points[:, 0] >= 0) & (points[:, 0] <= 10) & (points[:, 1] >= 0) & (points[:, 1] <= 10))]
fig, ax = plt.subplots(figsize=(6, 5))
ax.scatter(points[:, 0], points[:, 1], alpha=.7)
ax.set(xlim=(0, 10), ylim=(0, 10), title=f'Synthetic point pattern ({len(points)} events)', xlabel='x', ylabel='y')
ax.set_aspect('equal'); ax.grid(True); plt.show()
```

## 2. Compare bandwidths

The Gaussian kernel is fixed; only the bandwidth changes. A small bandwidth retains narrow peaks. A large bandwidth spreads each event's contribution farther across the window.

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4), sharex=True, sharey=True)
for ax, bandwidth in zip(axes, [0.35, 0.75, 1.5]):
    density, xs, ys = gaussian_kde(points, bounds=(0, 10, 0, 10), grid_shape=(90, 90), bandwidth=bandwidth)
    contour = ax.contourf(xs, ys, density, levels=16, cmap='viridis')
    ax.scatter(points[:, 0], points[:, 1], s=5, c='white', alpha=.5)
    ax.set_title(f'Bandwidth = {bandwidth}'); ax.set_aspect('equal')
    fig.colorbar(contour, ax=ax, shrink=.8, label='Intensity per map-unit²')
plt.tight_layout(); plt.show()
```

## 3. Add event weights

Weights change the contribution of each event. The resulting intensity is no longer simply event count per area; its units follow the weights.

```python
weights = np.ones(len(points))
weights[:10] = 3
weighted_density, xs, ys = gaussian_kde(points, bounds=(0, 10, 0, 10), grid_shape=(90, 90), bandwidth=.75, weights=weights)
fig, ax = plt.subplots(figsize=(6, 5))
im = ax.imshow(weighted_density, extent=(0, 10, 0, 10), origin='lower', cmap='magma')
ax.scatter(points[:, 0], points[:, 1], s=8, c='cyan', alpha=.6)
ax.set_title('Weighted event intensity'); ax.set_aspect('equal'); fig.colorbar(im, ax=ax, label='Weighted intensity'); plt.show()
```

## 4. Adaptive bandwidths and leave-one-out selection

Adaptive KDE varies the kernel bandwidth by event using a pilot density. Dense pilot regions receive smaller kernels, while sparse regions receive larger ones. The LOO score compares candidate global bandwidths by predicting each event from all other events.

```python
adaptive_density, xs, ys, local_h = adaptive_gaussian_kde(
    points, bounds=(0, 10, 0, 10), grid_shape=(90, 90), bandwidth=0.75,
    alpha=0.5, bandwidth_factor_bounds=(0.25, 4.0),
)
fig, axes = plt.subplots(1, 2, figsize=(11, 4), sharex=True, sharey=True)
fixed_density, _, _ = gaussian_kde(points, bounds=(0, 10, 0, 10), grid_shape=(90, 90), bandwidth=0.75)
for ax, surface, title in zip(axes, [fixed_density, adaptive_density], ['Fixed bandwidth', 'Adaptive sample-point bandwidth']):
    image = ax.imshow(surface, extent=(0, 10, 0, 10), origin='lower', cmap='viridis')
    ax.scatter(points[:, 0], points[:, 1], s=5, c='white', alpha=.35)
    ax.set(title=title, xlim=(0, 10), ylim=(0, 10)); ax.set_aspect('equal')
    fig.colorbar(image, ax=ax, label='Intensity')
plt.show()
print('Local bandwidth range:', float(local_h.min()), 'to', float(local_h.max()))
best_h, loo_scores = select_bandwidth_loo(points, [0.25, 0.4, 0.6, 0.75, 1.0, 1.4])
print('LOO-selected candidate:', best_h)
print('Candidate scores:', loo_scores)
```

LOO selection is only as useful as its candidate grid and sampling assumptions. It does not correct boundary bias. The adaptive estimator uses an event-level pilot and should be compared with the fixed-bandwidth surface rather than assumed to be universally superior.

## 5. Use a normal-reference bandwidth

Silverman's bivariate rule gives a reproducible starting scale from the pooled coordinate spread and effective sample size. Compare it with the LOO candidate-grid choice; neither method corrects boundary bias.


```python
h_silverman = silverman_bandwidth(points)
candidate_grid = [0.5 * h_silverman, 0.75 * h_silverman, h_silverman, 1.5 * h_silverman, 2 * h_silverman]
h_loo, scores = select_bandwidth_loo(points, candidate_grid)
print(f'Silverman normal-reference h: {h_silverman:.3f}')
print(f'LOO-selected h on supplied grid: {h_loo:.3f}')
print('Candidate mean log scores:', scores)
```
