# Weighted head-banging spatial smoothing

A small synthetic point pattern illustrates nearly collinear triples, low/high screens and robust smoothing of a local spike.

```python
import sys
sys.path.insert(0, "implementations/python/head-banging")
from head_bang import head_bang
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Create a synthetic surface with a spike

The broad gradient is retained while one isolated high value is supplied to the smoother.

```python
grid = [(float(x), float(y)) for y in range(5) for x in range(5)]
values = [float(2*x + y) for x,y in grid]
values[12] += 30
weights = [1.0] * len(values)
print('raw centre value:', values[12])
```

## 2. Apply weighted head-banging

The geometry is fixed, and all locations are updated synchronously for up to ten iterations.

```python
result = head_bang(grid, values, weights, n_neighbors=12, max_triples=8, max_iterations=10)
smoothed = result['smoothed_values']
print('smoothed centre value:', smoothed[12])
print('iterations:', result['iterations'], 'converged:', result['converged'])
raw = np.asarray(values).reshape(5,5)
smooth_grid = np.asarray(smoothed).reshape(5,5)
fig, axes = plt.subplots(1,2,figsize=(10,4))
axes[0].imshow(raw, origin='lower'); axes[0].set_title('Raw values')
axes[1].imshow(smooth_grid, origin='lower'); axes[1].set_title('Head-banged values')
for ax in axes: ax.set_xlabel('x'); ax.set_ylabel('y')
plt.tight_layout(); plt.show()
```

## 3. Interpretation

Head-banging is a robust descriptive smoother, not a risk model. The result depends on neighbour count, triple geometry and weights. Locations without qualifying triples remain unchanged in this implementation; the full NCI software has additional corner-point rules.
