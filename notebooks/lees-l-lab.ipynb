# Lee's L lab
Lee's L combines the spatially smoothed patterns of two variables. This is not the same as bivariate Moran's I.

```python
import sys
sys.path.insert(0, '../implementations/python/lees-l')
from lees_l import lees_l
```

## 1. Define a neighbourhood system
Four units form a ring. Each row is row-standardized by the implementation.

```python
weights = [
    [0, 1, 0, 0],
    [1, 0, 1, 0],
    [0, 1, 0, 1],
    [0, 0, 1, 0],
]
x = [1, 2, 4, 3]
y = [1, 2, 4, 3]
result = lees_l(x, y, weights)
print('Lee L:', round(result['L'], 4))
print('SSS x:', round(result['spatial_smoothing_scalar_x'], 4))
print('SSS y:', round(result['spatial_smoothing_scalar_y'], 4))
print('Spatial lag x:', result['spatial_lag_x'])
```

## 2. Compare different spatial patterns
The raw variable pairs can be changed while keeping the same graph. Observe how the spatially lagged values change, then compare Lee's L.

```python
reversed_pattern = lees_l(x, [4, 3, 1, 2], weights)
print('Same pattern:', round(result['L'], 4))
print('Reversed smoothed pattern:', round(reversed_pattern['L'], 4))
```

## 3. Interpretation and limits
Lee's L is global and symmetric in the two variables. It is not a p-value. The result depends on the neighbourhood graph and row-standardization. This teaching implementation rejects islands and does not calculate local L or permutation inference.
