# Lucas–Kanade optical flow lab

This notebook demonstrates dense local least-squares motion estimation for a small synthetic translation. It is not a full multi-scale optical-flow system.

```python
import sys
sys.path.insert(0, "implementations/python/optical-flow")
import math
import matplotlib.pyplot as plt
from lucas_kanade import lucas_kanade_dense

def gaussian(rows=31, cols=31, dx=0.0):
    return [[math.exp(-((j-dx-15)**2 + (i-15)**2)/35) for j in range(cols)] for i in range(rows)]

first = gaussian()
second = gaussian(dx=0.3)
flow = lucas_kanade_dense(first, second, window_radius=3)
print('valid pixels:', sum(sum(row) for row in flow['valid']))
```

```python
plt.figure(figsize=(6, 5))
plt.imshow(first, cmap='gray')
rows, cols = len(first), len(first[0])
x = [j for i in range(3, rows-3, 3) for j in range(3, cols-3, 3) if flow['valid'][i][j]]
y = [i for i in range(3, rows-3, 3) for j in range(3, cols-3, 3) if flow['valid'][i][j]]
u = [flow['u'][i][j] for i in range(3, rows-3, 3) for j in range(3, cols-3, 3) if flow['valid'][i][j]]
v = [flow['v'][i][j] for i in range(3, rows-3, 3) for j in range(3, cols-3, 3) if flow['valid'][i][j]]
plt.quiver(x, y, u, v, color='red'); plt.title('Estimated displacement (pixels/frame)'); plt.gca().invert_yaxis(); plt.show()
```
