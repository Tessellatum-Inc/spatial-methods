# Tucker decomposition lab: HOSVD and HOOI

Build a low multilinear-rank tensor, compute a truncated HOSVD approximation, then refine it with HOOI.

```python
import sys
sys.path.insert(0, "../implementations/python/tensor-decomposition")
import numpy as np
from tucker_hosvd import tucker_hosvd, tucker_hooi, tucker_reconstruct
```

```python
rng = np.random.default_rng(7)
A = rng.normal(size=(8, 2))
B = rng.normal(size=(6, 3))
C = rng.normal(size=(5, 2))
G = rng.normal(size=(2, 3, 2))
X = tucker_reconstruct(G, [A, B, C])
hosvd = tucker_hosvd(X, ranks=(2, 3, 2))
hooi = tucker_hooi(X, ranks=(2, 3, 2))
print('core shape:', hooi['core'].shape)
print('HOSVD relative error:', hosvd['relative_error'])
print('HOOI relative error:', hooi['relative_error'])
print('HOOI sweeps:', hooi['iterations'])
```

## Interpretation and limits
Try smaller ranks and compare error histories. HOOI is a local alternating refinement, not a guarantee of the global best Tucker fit. Do not zero-fill missing observations and treat them as measured data.
