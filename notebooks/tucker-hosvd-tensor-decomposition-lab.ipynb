# Tucker decomposition lab: truncated HOSVD

Build a low multilinear-rank tensor, compute its Tucker/HOSVD approximation, and inspect the core and reconstruction error.

```python
import numpy as np
from tucker_hosvd import tucker_hosvd, tucker_reconstruct
```

```python
rng = np.random.default_rng(7)
A = rng.normal(size=(8, 2))
B = rng.normal(size=(6, 3))
C = rng.normal(size=(5, 2))
G = rng.normal(size=(2, 3, 2))
X = tucker_reconstruct(G, [A, B, C])
fit = tucker_hosvd(X, ranks=(2, 3, 2))
print('core shape:', fit['core'].shape)
print('factor shapes:', [f.shape for f in fit['factors']])
print('relative reconstruction error:', fit['relative_error'])
```

## Compare truncation choices
Try smaller ranks and compare reconstruction error. Truncated HOSVD is a subspace projection, not generally the optimal Tucker fit. For iterative refinement, study HOOI separately. Do not zero-fill missing observations and treat them as measured data.
