# CP/PARAFAC decomposition lab

Construct a synthetic rank-one three-way tensor, fit CP-ALS, and inspect reconstruction error.

```python
import sys
sys.path.insert(0, "../implementations/python/tensor-decomposition")
import numpy as np
from cp_als import cp_als
```

```python
a = np.array([1., 2., 3., 4.])
b = np.array([2., -1., 3.])
c = np.array([1., .5])
X = np.einsum('i,j,k->ijk', a, b, c)
fit = cp_als(X, rank=1, max_iter=500, tolerance=1e-10, seed=3)
A, B, C = fit['factors']
X_hat = np.einsum('r,ir,jr,kr->ijk', fit['weights'], A, B, C)
print('relative reconstruction error:', np.linalg.norm(X-X_hat)/np.linalg.norm(X))
print('iterations:', fit['iterations'])
```

## Interpretation
A low reconstruction error on this synthetic example is not evidence that a decomposition of real spatiotemporal data is identifiable or scientifically meaningful. Explore higher rank, noise, and different initializations.
