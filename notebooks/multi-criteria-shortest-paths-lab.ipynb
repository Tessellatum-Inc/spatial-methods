# Pareto multi-criteria shortest paths

A single shortest-path cost cannot express every route trade-off. This lab finds the complete set of nondominated paths when each edge has a vector of nonnegative costs, such as travel time and distance.

```python
import sys
sys.path.insert(0, "implementations/python/shortest-paths")
from pareto_paths import pareto_shortest_paths
```

## 1. Time–distance trade-offs

Each edge stores (time, distance). The algorithm returns every nondominated source-to-target path; it does not choose a single compromise.

```python
graph = {
    's': [('a', (1, 8)), ('b', (7, 1)), ('c', (4, 4)), ('t', (12, 12))],
    'a': [('t', (1, 2))],
    'b': [('t', (3, 1))],
    'c': [('t', (4, 4))],
    't': [],
}
frontier = pareto_shortest_paths(graph, 's', 't', n_objectives=2)
for route in frontier:
    print(route)
```

## 2. More than two criteria

The implementation accepts any fixed number of additive nonnegative objectives. Dominance is componentwise, not a weighted sum.

```python
graph3 = {
    's': [('a', (0, 1, 2)), ('b', (1, 0, 2)), ('c', (1, 2, 0))],
    'a': [('t', (1, 0, 1))],
    'b': [('t', (0, 2, 0))],
    'c': [('t', (1, 0, 1))],
    't': [],
}
frontier3 = pareto_shortest_paths(graph3, 's', 't', n_objectives=3)
print([route['costs'] for route in frontier3])
```

## 3. Interpretation and limits

A Pareto frontier may contain many routes; the algorithm is exact but its worst-case output can be exponential. Edge costs must be nonnegative. The optional per-node label cap raises an error rather than silently truncating results. This is not a route-choice model, a transit network design model, or a multimodal timetable solver.
