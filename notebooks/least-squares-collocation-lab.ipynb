# Least-squares collocation lab
Combine a deterministic trend with a spatially correlated random signal and observation noise. All covariance matrices below are synthetic and supplied rather than estimated.

```python
import sys
sys.path.insert(0, "implementations/python/geostatistics")
import numpy as np
from least_squares_collocation import least_squares_collocation
```

```python
x=np.array([0.,1.,2.,3.,4.])
d=np.abs(x[:,None]-x[None,:])
C=np.exp(-d/1.5)
R=np.eye(len(x))*0.15
H=np.ones((len(x),1))
y=5.0+np.array([.2,-.1,.05,.15,-.05])
xt=np.array([.5,1.5,2.5,3.5])
d0=np.abs(xt[:,None]-x[None,:])
C0=np.exp(-d0/1.5)
C00=np.exp(-np.abs(xt[:,None]-xt[None,:])/1.5)
result=least_squares_collocation(y,H,C,R,C0,C00,np.ones((len(xt),1)))
print('trend estimate:',result['trend_parameters'])
print('predictions:',np.round(result['prediction'],3))
print('prediction standard deviations:',np.round(result['prediction_standard_deviation'],3))
```

## Interpretation
Prediction depends on the supplied signal covariance, measurement-noise covariance and trend design. Poor covariance assumptions can produce misleading precision. This dense example is a learning baseline, not a large-scale gravity-field or production interpolation system.
