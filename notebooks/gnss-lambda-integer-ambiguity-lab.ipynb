# GNSS LAMBDA integer ambiguity lab
A synthetic float ambiguity vector and covariance are decorrelated with an integer-unimodular transformation, searched in the reduced domain, then mapped back. This is an ILS solver, not a raw GNSS estimator or PPP-AR engine.

```python
import sys
sys.path.insert(0, '../implementations/python/gnss-ambiguity')
import numpy as np
from lambda_ils import lambda_reduction, lambda_integer_least_squares
```

```python
a = np.array([1.2, 1.8])
Q = np.array([[4.0, 3.9], [3.9, 4.0]])
reduced = lambda_reduction(a, Q)
reduced['transformation'], reduced['transformed_ambiguities'], reduced['transformed_covariance']
```

```python
solution = lambda_integer_least_squares(a, Q)
{key: solution[key] for key in ('best', 'best_objective', 'second_best', 'second_best_objective', 'ratio', 'visited_nodes')}
```

## Interpretation
The transform is integer-unimodular, so it preserves the integer lattice and the original quadratic objective. The second-best/best objective ratio is diagnostic only; it is not an automatic acceptance decision.
