# Gravity models with PPML
A synthetic example retaining zero flows. This is not a replication of published tables.

```python
import sys
sys.path.insert(0, "../implementations/python/regional-dynamics")
import numpy as np
from ppml import ppml
rng=np.random.default_rng(2026)
n=2500
z=rng.normal(2.0,0.5,n)
lo=rng.normal(8.0,0.7,n)
ld=rng.normal(8.0,0.7,n)
x=np.column_stack([np.ones(n),z,lo,ld])
beta=np.array([0.2,-0.6,0.7,0.65])
y=rng.poisson(np.exp(x@beta))
fit=ppml(y,x)
print("zero share:",np.mean(y==0))
print("coefficients:",fit.coefficients)
print("HC0 standard errors:",fit.standard_errors)
print("converged / iterations:",fit.converged,fit.iterations)
```

## Interpretation
The logged-distance coefficient is an elasticity. PPML permits zero flows and models the conditional mean. HC0 is heteroskedasticity-robust but does not account for dyadic or panel clustering.
