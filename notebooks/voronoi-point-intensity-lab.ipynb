# Voronoi intensity estimation lab
Construct clipped Voronoi cells in a bounded rectangular window and inspect the reciprocal-area intensity estimate.

```python
import sys
sys.path.insert(0, "implementations/python/spatial-intensity")
import numpy as np
from voronoi_intensity import voronoi_intensity
```

```python
points=np.array([[1,1],[2,7],[4,3],[6,8],[8,2],[9,6]],dtype=float)
result=voronoi_intensity(points,[0,0,10,10])
for i,(area,intensity,cell) in enumerate(zip(result['areas'],result['intensity'],result['cells'])):
    print(i,'area=',round(area,3),'intensity=',round(intensity,4),'vertices=',len(cell))
print('partition area:',result['areas'].sum())
```

## Interpretation
The estimate is piecewise constant per cell. Small clipped cells yield large local intensity values. Since cells are clipped to the observation window and no exterior guard points or edge correction are used, boundary estimates can be biased. This raw estimator is also more variable than smoothed alternatives.
