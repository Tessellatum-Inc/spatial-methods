# Residual-based GNSS RAIM lab
A synthetic linearised snapshot: global chi-square detection followed by single-measurement leave-one-out exclusion. This is not certified RAIM and does not calculate protection levels.

```python
import sys
sys.path.insert(0, '../implementations/python/gnss-raim')
import numpy as np
from raim import residual_global_test, single_fault_exclusion
```

```python
epochs = np.arange(7.0)
H = np.column_stack([np.ones_like(epochs), epochs])
y = 12.0 + 0.8 * epochs + np.array([.01, -.02, .01, 0, -.01, .02, .0])
y[5] += 8.0  # injected pseudorange-like fault
R = np.eye(len(y)) * 0.04
global_result = residual_global_test(y, H, R, false_alarm_probability=0.01)
global_result['alarm'], global_result['weighted_residual_sum_squares'], global_result['threshold']
```

```python
fde = single_fault_exclusion(y, H, R, false_alarm_probability=0.01)
fde['status'], fde['excluded_index'], fde['candidate_tests']
```

## Interpretation
A global alarm means the observation set is inconsistent with the supplied linearised model and covariance. A unique passing deletion is only a diagnostic candidate under a single-fault hypothesis. Multiple candidates remain ambiguous. No protection level, availability, continuity or aviation integrity guarantee is produced.
