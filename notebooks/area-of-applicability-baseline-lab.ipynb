# Area-of-applicability baseline lab
A nearest-neighbour predictor-space envelope. This is a transparent teaching baseline, not an exact Meyer–Pebesma AOA implementation.

```python
import sys, numpy as np
sys.path.insert(0, '../implementations/python/distribution-shift')
from area_of_applicability import fit_applicability_envelope, predict_applicability
```

```python
rng=np.random.default_rng(12)
train=rng.normal(size=(120,3))
model=fit_applicability_envelope(train, quantile=.95)
target=np.vstack([rng.normal(size=(40,3)), rng.normal(loc=4, size=(20,3))])
result=predict_applicability(model,target)
print('threshold:',model['threshold'])
print('applicable fraction:',result['applicable'].mean())
print('outside count:',np.count_nonzero(~result['applicable']))
```

```python
import matplotlib.pyplot as plt
plt.scatter(target[:,0],target[:,1],c=result['applicable'],s=22)
plt.xlabel('Predictor 1'); plt.ylabel('Predictor 2')
plt.title('Applicability flags (displaying first two predictors)'); plt.show()
```

## Important limitations
The threshold is a quantile of leave-one-out nearest-neighbour distances. This baseline uses training feature ranges and optional user-supplied feature weights. It is not a calibrated prediction-error bound, does not estimate variable importance from a fitted model, and does not account for spatial dependence in the threshold sample. Select the validation design and threshold before examining held-out target outcomes.
