# Regional beta-convergence lab
Fit an absolute beta-convergence regression to a synthetic regional cross-section, then compare it with a conditional model including a control.

```python
import sys
sys.path.insert(0, "../implementations/python/regional-dynamics")
import numpy as np
from beta_convergence import beta_convergence
```

```python
rng=np.random.default_rng(22)
y0=np.exp(np.linspace(1.5,3.5,50))
control=rng.normal(size=50)
growth=.07-.14*np.log(y0)+.02*control+rng.normal(0,.004,size=50)
y1=y0*np.exp(growth*10)
absolute=beta_convergence(y0,y1,periods=10)
conditional=beta_convergence(y0,y1,periods=10,controls=control)
print('absolute beta:',absolute['beta'],'half-life:',absolute['half_life_periods'])
print('conditional beta:',conditional['beta'],'HC1 SE:',conditional['hc1_standard_errors'])
```

## Interpretation
A negative initial-income coefficient is consistent with beta-convergence under the fitted specification. The conditional regression changes the comparison by controlling for an observed covariate. Neither model accounts for spatially correlated errors, endogenous regressors or omitted spatial interactions; the standard errors do not solve those problems.
