# Graph-fused spatial regression lab
Fit location-specific coefficients with sparsity and graph total-variation penalties.

```python
import sys
sys.path.insert(0, "../implementations/python/spatial-regression")
import numpy as np
from graph_fused_lasso import graph_fused_lasso
rng = np.random.default_rng(7)
nloc = 5
locations = np.repeat(np.arange(nloc), 100)
X = rng.normal(size=(len(locations), 2))
beta_true = np.array([[1.0,0.0],[1.05,0.0],[1.1,0.0],[2.0,0.5],[2.05,0.5]])
y = np.sum(X * beta_true[locations], axis=1) + rng.normal(0,0.15,len(locations))
edges = [(0,1),(1,2),(2,3),(3,4)]
```

```python
fit = graph_fused_lasso(X,y,locations,nloc,edges,lambda_l1=0.08,lambda_fused=2.0)
fit['coefficients'], fit['converged'], fit['iterations']
```

```python
# Compare a nearly pooled fit with weak and strong graph fusion.
weak = graph_fused_lasso(X,y,locations,nloc,edges,lambda_l1=0.08,lambda_fused=0.01)
strong = graph_fused_lasso(X,y,locations,nloc,edges,lambda_l1=0.08,lambda_fused=20.0)
weak['coefficients'], strong['coefficients']
```

## Interpretation
The graph is a modeling assumption. Validate its construction and tune penalties with spatially structured holdouts; random row splits can leak information between neighboring locations.
