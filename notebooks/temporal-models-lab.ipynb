## Decode a simple hidden-state sequence
The HMM parameters are supplied, not learned. State labels are arbitrary and must be interpreted using the model's emission/transition design.

```python
observations = [0, 0, 1, 1, 0]
start = [0.6, 0.4]
transition = [[0.9, 0.1], [0.1, 0.9]]
emission = [[0.9, 0.1], [0.2, 0.8]]
viterbi(observations, start, transition, emission)
```

## Interpretation safeguards
- Theil–Sen does not remove seasonality or serial dependence.
- Pettitt is a single-change test; its p-value approximation is not a spatial multiple-testing correction.
- Viterbi returns the most probable path conditional on the supplied HMM, not proof that hidden states are physical classes.
