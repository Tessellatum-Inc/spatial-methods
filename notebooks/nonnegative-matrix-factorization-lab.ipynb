# Non-negative matrix factorization lab

Fit a low-rank non-negative factorization to a small matrix. The example is synthetic and demonstrates numerical behavior, not a spatial interpretation.

```python
import sys
sys.path.insert(0, "implementations/python/matrix-factorization")
from nmf import nmf_euclidean

V = [[1, 2, 3, 4], [2, 4, 6, 8], [3, 6, 9, 12]]
fit = nmf_euclidean(V, rank=1, max_iter=500)
print('Final reconstruction error:', fit['errors'][-1])
print('W:', fit['W'])
print('H:', fit['H'])
```

```python
import matplotlib.pyplot as plt
plt.plot(fit['errors']); plt.xlabel('Iteration'); plt.ylabel('Frobenius error'); plt.title('NMF objective history'); plt.show()
```
