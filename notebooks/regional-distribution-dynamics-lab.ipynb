# Regional distribution dynamics lab

Compare ordinary Markov transitions, spatially conditioned transitions, LISA quadrant mobility and sigma-dispersion. Synthetic balanced panel only; not an empirical regional study.

```python
import sys
sys.path.insert(0, "../implementations/python/regional-dynamics")
import numpy as np
from regional_dynamics import markov_transitions, spatial_markov_transitions, lisa_transition_matrix, sigma_convergence
```

```python
# Eight regions arranged in a ring.
n = 8
W = np.zeros((n, n))
for i in range(n):
    W[i, (i-1) % n] = 1
    W[i, (i+1) % n] = 1
base = np.array([1., 1.4, 1.8, 2.2, 4.5, 4.9, 5.3, 5.7])
panel = np.vstack([base, base * 1.05 + np.array([.1,0,0,.1,-.1,0,.1,0]), base * 1.1])
ordinary = markov_transitions(panel, n_classes=3)
spatial = spatial_markov_transitions(panel, W, n_classes=3)
lisa = lisa_transition_matrix(panel, W)
print('ordinary counts\n', ordinary['counts'])
print('ordinary probabilities\n', np.round(ordinary['probabilities'], 3))
print('spatial conditional tensor shape', spatial['probabilities'].shape)
print('LISA labels', lisa['labels'])
print('LISA transitions\n', lisa['counts'])
print('sigma by period', sigma_convergence(panel))
```
