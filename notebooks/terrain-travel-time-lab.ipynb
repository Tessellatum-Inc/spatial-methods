# Directional terrain travel time

Compare Tobler's slope-dependent speed with Naismith and Langmuir rule-based travel times on a tiny elevation raster. Times are accumulated in seconds; the result is directional.

```python
import sys
sys.path.insert(0, "implementations/python/cost-distance")
import numpy as np
from terrain_travel import least_cost_time, tobler_speed_kmh, naismith_time_seconds, langmuir_time_seconds

elevation = np.array([[100., 110., 125.], [98., 105., 118.], [95., 101., 110.]])
source = [(0, 0)]
```

```python
for model in ("tobler", "naismith", "langmuir"):
    travel = least_cost_time(elevation, source, model=model, cell_size_x=100, cell_size_y=100)
    print(model, "seconds to far corner:", travel[-1, -1])

print("Tobler flat speed km/h:", tobler_speed_kmh(0.0))
print("Naismith 5 km flat seconds:", naismith_time_seconds(5000, 0))
print("Langmuir 300 m moderate descent seconds:", langmuir_time_seconds(5000, 0, 300, 10))
```

```python
uphill = least_cost_time([[0., 10.]], [(0, 0)], model="tobler", cell_size_x=100, connectivity=4)
downhill = least_cost_time([[0., 10.]], [(0, 1)], model="tobler", cell_size_x=100, connectivity=4)
assert uphill[0, 1] > downhill[0, 0]
assert naismith_time_seconds(5000, 0) == 3600
print("Directional and rule-rate checks passed.")
```
