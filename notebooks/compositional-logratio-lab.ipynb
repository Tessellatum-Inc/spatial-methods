# Compositional data and log-ratio transforms

A composition describes relative parts of a whole. This lab applies closure, CLR, ALR, ILR and Aitchison distance while making the zero-value limitation explicit.

```python
from pathlib import Path
import sys
import numpy as np
ROOT = Path.cwd()
if not (ROOT / 'implementations/python/compositional-data/logratio.py').is_file():
    ROOT = ROOT.parent
MODULE_DIR = ROOT / 'implementations/python/compositional-data'
if not (MODULE_DIR / 'logratio.py').is_file():
    raise FileNotFoundError(f'Cannot locate compositional-data implementation: {MODULE_DIR}')
sys.path.insert(0, str(MODULE_DIR))
from logratio import closure, clr, ilr, ilr_inverse, alr, alr_inverse, aitchison_distance
```

## 1. Closure and scale invariance

The total is arbitrary for compositional data. Closure maps the parts to a unit-sum composition.

```python
parts = np.array([12., 25., 63.])
print('closed:', closure(parts))
print('same after rescaling:', closure(parts * 1000.))
```

## 2. CLR coordinates

CLR coordinates are centred log ratios and sum to zero. The full D-dimensional covariance is singular because of this constraint.

```python
print('CLR:', clr(parts))
print('CLR sum:', clr(parts).sum())
```

## 3. ILR coordinates and inverse

The Helmert basis maps the D-part composition to D−1 orthonormal coordinates.

```python
coordinates = ilr(parts)
reconstructed = ilr_inverse(coordinates)
print('ILR:', coordinates)
print('inverse:', reconstructed)
print('round-trip error:', np.max(np.abs(reconstructed - closure(parts))))
```

## 4. ALR and Aitchison distance

ALR uses a selected denominator. Aitchison distance is invariant to the choice of ILR basis.

```python
print('ALR, last part denominator:', alr(parts))
print('ALR inverse:', alr_inverse(alr(parts)))
other = np.array([20., 20., 60.])
print('Aitchison distance:', aitchison_distance(parts, other))
```

## 5. Zeros are a modelling decision

A log-ratio cannot be evaluated at zero. Do not silently add a pseudocount: first decide whether a zero is structural, rounded or below a detection limit. This implementation rejects zeros in log-ratio methods and leaves replacement to a justified domain model.
