# Third-order tensor SVD (t-SVD)
Explore the FFT-based t-SVD and tubal-rank truncation. This is distinct from CP, Tucker and tensor-train decompositions.

```python
import sys
sys.path.insert(0, "implementations/python/tensor-decomposition")
import numpy as np
from t_svd import t_svd, t_svd_reconstruct, t_svd_truncate

rng=np.random.default_rng(2026)
x=rng.normal(size=(12,9,7))
full=t_svd_reconstruct(*t_svd(x))
print("full reconstruction error:",np.linalg.norm(x-full))
for rank in [1,2,4,7,9]:
    approx,_=t_svd_truncate(x,rank)
    print(rank,np.linalg.norm(x-approx)/np.linalg.norm(x))
```

## Interpretation
The rank parameter is tubal rank under the t-product, not CP rank or Tucker multilinear rank. FFT imposes circular convolution along the third mode; consider whether that boundary assumption is meaningful for the application.
