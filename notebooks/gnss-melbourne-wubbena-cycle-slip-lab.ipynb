Compute the wide-lane observable from carrier phase cycles and code ranges, then screen a synthetic wide-lane jump.

```python
import sys
from pathlib import Path
cwd = Path.cwd().resolve()
repo_root = next((p for p in (cwd, *cwd.parents) if (p / "implementations/python/gnss-ambiguity").is_dir()), None)
if repo_root is None:
    raise RuntimeError("Could not locate repository root from " + str(cwd))
sys.path.insert(0, str(repo_root / "implementations/python/gnss-ambiguity"))
from combinations import melbourne_wubbena
from melbourne_wubbena import detect_melbourne_wubbena_slips
import numpy as np
f1, f2 = 1575.42e6, 1227.60e6
```

```python
n = 40
l1 = 1000 + np.arange(n) * 0.01
l2 = 800 + np.arange(n) * 0.01
p1 = np.full(n, 21_000_000.0)
p2 = np.full(n, 21_000_000.0)
mw = melbourne_wubbena(l1, l2, p1, p2, f1, f2)
mw[:3]
```

```python
mw = np.asarray(mw)
mw[20:] += 1.0
screen = detect_melbourne_wubbena_slips(mw, initial_sigma=0.1, sigma_multiplier=4.0)
np.flatnonzero(screen['flags']), screen['thresholds'][20]
```

## Interpretation
