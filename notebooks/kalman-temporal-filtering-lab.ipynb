# Kalman local-level filter and RTS smoother
The filter estimates a latent random-walk level causally. The RTS smoother revises earlier estimates using later observations.

```python
import sys
sys.path.insert(0, "implementations/python/temporal-models")
from pathlib import Path
import sys
sys.path.insert(0, str(Path.cwd() / 'implementations/python/temporal-models'))
from kalman import kalman_local_level, rts_local_level
observations = [1.0, 1.1, None, 0.9, 1.2, 1.0]
filtered_mean, filtered_var = kalman_local_level(
    observations, process_variance=0.02, observation_variance=0.1
)
smoothed_mean, smoothed_var = rts_local_level(
    observations, process_variance=0.02, observation_variance=0.1
)
list(zip(filtered_mean, smoothed_mean))
```

## Interpretation
The filter output uses observations only up to each time. The RTS output uses the full series and is therefore retrospective. Tune process and observation variance using a defensible measurement model; smoothing may erase abrupt events.
