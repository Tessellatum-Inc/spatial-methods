# Moran eigenvector spatial filtering

Build an orthogonal spatial basis from a symmetric weights matrix and compare OLS with a fixed-size eigenvector-augmented regression.

```python
import sys
sys.path.insert(0, "implementations/python/moran-eigenvector-filtering")
from moran_filter import moran_eigenvectors, moran_eigenvector_filter
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Build a simple symmetric weights matrix

A path graph is used as a small illustrative spatial connectivity structure.

```python
n = 20
W = np.zeros((n,n))
for i in range(n-1): W[i,i+1] = W[i+1,i] = 1.0
basis = moran_eigenvectors(W, n_components=3)
print('leading eigenvalues:', basis['eigenvalues'])
print('basis orthogonality error:', np.max(np.abs(basis['eigenvectors'].T @ basis['eigenvectors'] - np.eye(3))))
```

## 2. Fit a synthetic spatial signal

The number of eigenvectors is supplied by the analyst. This lab does not implement an automatic selection criterion.

```python
rng = np.random.default_rng(7)
x = np.linspace(0,1,n)
y = 3 + 2*x + 2.5*basis['eigenvectors'][:,0] + rng.normal(0,0.25,n)
ols = moran_eigenvector_filter(y, x, W, n_components=0)
filtered = moran_eigenvector_filter(y, x, W, n_components=2)
print('OLS residual sum of squares:', ols['residual_sum_squares'])
print('filtered residual sum of squares:', filtered['residual_sum_squares'])
plt.plot(x,y,'o',label='observed')
plt.plot(x,ols['fitted'],label='OLS')
plt.plot(x,filtered['fitted'],label='eigenvector filter')
plt.legend(); plt.xlabel('covariate'); plt.ylabel('response'); plt.show()
```

## 3. Interpretation

The eigenvectors are synthetic spatial patterns, not observed covariates. Adding more can only reduce in-sample residual sum of squares; it does not guarantee better out-of-sample prediction or valid inference. Eigenvector selection and residual spatial diagnostics remain essential.
