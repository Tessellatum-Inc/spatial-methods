# Cascading overload failures
A small Motter–Lai-style shortest-path load experiment. This is a topology model, not an engineering power-flow simulation.

```python
import sys
sys.path.insert(0, "implementations/python/network-resilience")
from cascading_failures import node_betweenness_load, motter_lai_cascade

graph={0:{1,3,4},1:{0,2},2:{1,3},3:{0,2},4:{0}}
print("initial node loads:",node_betweenness_load(graph))
result=motter_lai_cascade(graph,[1],tolerance=0.2)
print("failure rounds:",result["failure_rounds"])
print("cascade fraction:",result["cascade_fraction"])
print("survivors:",result["surviving_nodes"])
```

## Sensitivity to tolerance
Increase tolerance to represent greater overload headroom. The model assumes shortest-path rerouting and synchronous failures; it is not a calibrated representation of any particular road, power, water or communication network.

```python
for alpha in [0.0,0.2,0.5,1.0,2.0]:
    out=motter_lai_cascade(graph,[1],tolerance=alpha)
    print(alpha,out["cascade_size"],out["failure_rounds"])
```
