# Tukey median polish

Decompose a two-way table into overall, row and column effects with robust median-based updates.

```python
import sys
sys.path.insert(0, "implementations/python/median-polish")
from median_polish import median_polish
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Example two-way table

The table is a classic example used in R's medpolish documentation.

```python
table = [[14,15,14],[7,4,7],[8,2,10],[15,9,10],[0,2,0]]
result = median_polish(table)
print('overall:', result['overall'])
print('row effects:', result['row_effects'])
print('column effects:', result['column_effects'])
print('iterations:', result['iterations'], 'converged:', result['converged'])
```

## 2. Inspect the residual table

Large residuals are cell-specific departures from the additive row-plus-column pattern.

```python
resid = np.asarray(result['residuals'], dtype=float)
fig, ax = plt.subplots(figsize=(6,4))
im = ax.imshow(resid, cmap='coolwarm', aspect='auto')
ax.set_title('Median-polish residuals'); ax.set_xlabel('Column'); ax.set_ylabel('Row')
fig.colorbar(im, ax=ax, label='Residual'); plt.show()
```

## 3. Missing values

By default missing cells are rejected. With na_rm=True, medians ignore missing cells and their residual entries remain missing. All-missing rows or columns receive zero effects.

```python
with_missing = [[1,2,None],[3,4,5]]
missing_result = median_polish(with_missing, na_rm=True)
print(missing_result['residuals'])
```
