# Ensemble transform Kalman filter lab
Run a deterministic batch ETKF analysis and compare the ensemble analysis moments with the conventional Kalman update formed from the same finite-ensemble covariance.

```python
import sys
from pathlib import Path

# nbconvert executes notebooks from the notebook directory, not always the repository root.
repo_root = Path.cwd()
if not (repo_root / "implementations").is_dir():
    repo_root = repo_root.parent
module_dir = repo_root / "implementations" / "python" / "state-estimation"
if not module_dir.is_dir():
    raise RuntimeError(f"Could not locate state-estimation package from {Path.cwd()}")
sys.path.insert(0, str(module_dir))
import numpy as np
from ensemble_square_root import etkf_analysis
```

```python
rng=np.random.default_rng(18)
ensemble=rng.normal(size=(4,100))
H=np.array([[1.,0.,1.,0.],[0.,1.,0.,1.]])
R=np.array([[.3,.05],[.05,.5]])
y=np.array([1.0,-.4])
result=etkf_analysis(ensemble,y,H,R)
print('forecast mean:',np.round(result['forecast_mean'],3))
print('analysis mean:',np.round(result['analysis_mean'],3))
print('analysis standard deviations:',np.round(np.sqrt(np.diag(result['analysis_covariance'])),3))
print('ensemble mean consistency:',np.max(np.abs(result['analysis_ensemble'].mean(axis=1)-result['analysis_mean'])))
```

## Interpretation
The update is deterministic once the forecast ensemble and observations are fixed. The analysis covariance is confined to the forecast ensemble subspace; a small or poorly spread ensemble cannot represent uncertainty directions absent from that subspace. Localization and inflation are important operational extensions but are not included here.
