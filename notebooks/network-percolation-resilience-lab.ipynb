# Network percolation and resilience lab

Compare random node failures with static degree-targeted attacks. This notebook tracks topology only, not service or demand.

```python
import sys
sys.path.insert(0, "implementations/python/network-resilience")
from percolation import degree_targeted_order, random_attack_order, node_attack_curve

graph = {0:{1,2,3,4}, 1:{0,5}, 2:{0,5}, 3:{0}, 4:{0}, 5:{1,2,6}, 6:{5}}
targeted = node_attack_curve(graph, degree_targeted_order(graph))
randomized = node_attack_curve(graph, random_attack_order(graph, seed=17))
print('Targeted:', targeted)
print('Random:', randomized)
```

```python
import matplotlib.pyplot as plt
plt.plot(targeted['removed_fraction'], targeted['largest_component_fraction'], label='Degree targeted')
plt.plot(randomized['removed_fraction'], randomized['largest_component_fraction'], label='Random')
plt.xlabel('Fraction of original nodes removed'); plt.ylabel('Largest component / original nodes')
plt.legend(); plt.show()
```
