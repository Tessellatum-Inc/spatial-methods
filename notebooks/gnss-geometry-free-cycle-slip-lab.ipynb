# Geometry-free GNSS cycle-slip screening lab

Synthetic geometry-free phase series with a quadratic background trend and a simulated discontinuity. The detector flags candidate epochs only; it does not repair integer ambiguities.

```python
import sys
from pathlib import Path
cwd = Path.cwd().resolve()
repo_root = next((p for p in (cwd, *cwd.parents) if (p / "implementations/python/gnss-ambiguity").is_dir()), None)
if repo_root is None:
    raise RuntimeError("Could not locate repository root from " + str(cwd))
sys.path.insert(0, str(repo_root / "implementations/python/gnss-ambiguity"))
import numpy as np
from cycle_slips import detect_geometry_free_slips
```

```python
t = np.arange(40, dtype=float)
gf = 0.001*t**2 + 0.02*t
gf[24:] += 0.4  # synthetic phase discontinuity in metres
result = detect_geometry_free_slips(gf, epochs=t, window=7, threshold=0.08)
print('flagged epochs:', np.flatnonzero(result['flags']))
print('residual at simulated event:', result['residuals'][24])
```

## Cautions
- Increase ionospheric variability and measurement noise to explore false alarms.
- Vary the sampling interval and threshold; the threshold is not universal.
- A production dual-frequency detector should combine complementary GF and Melbourne–Wübbena evidence and handle slip identification/repair.
