# Spatial and space-time KDE lab

Estimate event intensity at query locations using planar and separable space-time Gaussian kernels.

```python
import sys
sys.path.insert(0, "implementations/python/spatial-intensity")
import numpy as np
from kde import gaussian_spatial_intensity, gaussian_spacetime_intensity
rng = np.random.default_rng(4)
events = rng.normal(size=(80, 2))
event_times = rng.uniform(0, 30, size=80)
queries = np.array([[0., 0.], [1., 1.], [-1., 0.5]])
query_times = np.array([10., 10., 20.])
```

```python
spatial = gaussian_spatial_intensity(events, queries, bandwidth=0.6)
spacetime = gaussian_spacetime_intensity(events, event_times, queries, query_times, 0.6, 4.0)
print('spatial intensity:', spatial)
print('space-time intensity:', spacetime)
```

## Interpretation
These are intensity estimates, not probabilities. Change the bandwidths and inspect the effect on smoothness. The estimators do not correct for study-window boundaries, varying observation effort, road-network constraints or anisotropy.
