# Constrained gravity and entropy-maximising flow models

A small synthetic origin-destination example compares unconstrained, singly constrained and doubly constrained flow allocation. The doubly constrained model uses iterative proportional fitting; the entropy wrapper uses exponential generalized-cost deterrence.

```python
import sys
sys.path.insert(0, "implementations/python/regional-dynamics")
import numpy as np
from gravity_flows import (unconstrained_gravity, production_constrained_gravity,
    attraction_constrained_gravity, doubly_constrained_gravity, entropy_maximizing_gravity)

origins = np.array([120., 80.])
destinations = np.array([60., 90., 50.])
impedance = np.array([[1.0, 0.6, 0.2], [0.4, 1.0, 0.8]])
cost = np.array([[2., 5., 12.], [8., 3., 4.]])
```

```python
free = unconstrained_gravity(origins, destinations, impedance)
production = production_constrained_gravity(origins, destinations, impedance)
attraction = attraction_constrained_gravity(origins, destinations, impedance)
balanced = doubly_constrained_gravity(origins, destinations, impedance)
entropy = entropy_maximizing_gravity(origins, destinations, cost, beta=0.15)

print("Unconstrained total:", free.sum())
print("Production row totals:", production.sum(axis=1))
print("Attraction column totals:", attraction.sum(axis=0))
print("Doubly constrained rows:", balanced.flows.sum(axis=1))
print("Doubly constrained columns:", balanced.flows.sum(axis=0))
print("Entropy-maximising convergence:", entropy.converged, entropy.iterations)
```

```python
assert np.allclose(production.sum(axis=1), origins)
assert np.allclose(attraction.sum(axis=0), destinations)
assert balanced.converged and entropy.converged
assert np.allclose(balanced.flows.sum(axis=1), origins, atol=1e-7)
assert np.allclose(balanced.flows.sum(axis=0), destinations, atol=1e-7)
assert np.allclose(entropy.flows.sum(axis=1), origins, atol=1e-7)
assert np.allclose(entropy.flows.sum(axis=0), destinations, atol=1e-7)
print("All margin constraints verified.")
```
