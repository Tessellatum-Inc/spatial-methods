# Globe-based mean radiant temperature
Estimate mean radiant temperature from globe temperature, air temperature and local air speed. This is a steady-state ISO 7726-style globe correction, not a complete UTCI/PET model.

```python
import sys
from pathlib import Path
root = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / 'implementations/python/mean-radiant-temperature').exists())
sys.path.insert(0, str(root / 'implementations/python/mean-radiant-temperature'))
from mean_radiant_temperature import mean_radiant_temperature
```

```python
# Standard 150 mm matte-black globe example
result = mean_radiant_temperature(
    globe_temperature_c=35.0,
    air_temperature_c=28.0,
    air_velocity_m_s=0.2,
)
result
```

## Compare convection regimes
The implementation calculates natural and forced convection coefficients and uses the larger one. Increase air speed to see when forced convection controls.

```python
for speed in (0.0, 0.1, 0.5, 1.0):
    estimate = mean_radiant_temperature(35.0, 28.0, speed)
    print(speed, estimate)
```
