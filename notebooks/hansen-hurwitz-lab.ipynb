# Hansen–Hurwitz: unequal-probability sampling with replacement

This lab uses a small finite population to show how the draw-by-draw inverse-probability estimator treats repeated selections. It demonstrates the estimator contract; it is not a sampling-design simulator or a confidence-interval lesson.

```python
from pathlib import Path
import sys
ROOT = Path.cwd()
if not (ROOT / 'implementations').exists(): ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'implementations/python/hansen-hurwitz'))
from hansen_hurwitz import hansen_hurwitz
```

## 1. The ordered draw sequence

The values and probabilities below represent the original draw sequence. A repeated unit remains repeated because Hansen–Hurwitz works on draws, not on a deduplicated set of selected units.

```python
values = [2, 4, 6, 4]
draw_probabilities = [0.2, 0.4, 0.2, 0.4]
result = hansen_hurwitz(values, draw_probabilities, population_size=10)
print('draw-level estimates:', result.draw_estimates)
print('estimated total:', result.total_estimate)
print('estimated total variance:', result.total_variance_estimate)
print('estimated population mean:', result.mean_estimate)
print('estimated mean variance:', result.mean_variance_estimate)
```

## 2. Why repeated draws are retained

The returned draw-level estimates are \(y_i/p_i\). The total estimate averages them. For independent with-replacement draws, the estimated variance of the total is the sample variance of these draw-level estimates divided by the number of draws. If only one draw is available, the sample variance cannot be estimated and the implementation returns `None`.

This is not the Horvitz–Thompson inclusion-probability estimator for sampling without replacement. The design and the probabilities must be known; these arrays alone do not establish an unbiased design.

```python
one_draw = hansen_hurwitz([5], [0.1], population_size=20)
print(one_draw.total_estimate, one_draw.total_variance_estimate, one_draw.mean_estimate)
```
