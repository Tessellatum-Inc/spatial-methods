# Louvain community detection lab
Build a weighted undirected graph with two dense blocks joined by a weak bridge. Run multilevel Louvain, inspect modularity and vary the resolution parameter. The method is heuristic; a seed controls traversal order, not a guarantee of a unique optimum.

```python
from pathlib import Path
import sys
sys.path.insert(0, str(Path('../implementations/python/louvain').resolve()))
from community import louvain_communities, modularity
```

```python
graph = {node: {} for node in range(8)}
for block in (range(4), range(4, 8)):
    for u in block:
        for v in block:
            if u < v:
                graph[u][v] = graph[v][u] = 1.0
graph[3][4] = graph[4][3] = 0.05
communities = louvain_communities(graph, seed=7)
communities
```

```python
print('Communities:', communities)
print('Modularity:', round(modularity(graph, communities), 4))
print('Singleton modularity:', round(modularity(graph, [{u} for u in graph]), 4))
assert {frozenset(c) for c in communities} == {frozenset(range(4)), frozenset(range(4, 8))}
```

```python
for gamma in (0.5, 1.0, 2.0):
    partition = louvain_communities(graph, resolution=gamma, seed=7)
    print(f'resolution={gamma}: {len(partition)} communities; Q={modularity(graph, partition, gamma):.4f}')
```
