## Interpretation
Monte Carlo uncertainty shrinks approximately with the square root of trial count, but direct sampling is weak for rare failures. Minimal cuts describe topological combinations, not their likelihood or consequence. Static betweenness is a ranking heuristic, not a service-impact model.
