# GNSS integer ambiguity search lab

This lab starts from an already-estimated float ambiguity vector and covariance. It does not process carrier-phase observations or perform LAMBDA decorrelation.

```python
import sys
from pathlib import Path
cwd = Path.cwd().resolve()
repo_root = next((p for p in (cwd, *cwd.parents) if (p / "implementations/python/gnss-ambiguity").is_dir()), None)
if repo_root is None:
    raise RuntimeError("Could not locate repository root from " + str(cwd))
sys.path.insert(0, str(repo_root / "implementations/python/gnss-ambiguity"))
from integer_least_squares import integer_least_squares

float_ambiguities = [12.18, -3.73]
covariance = [[0.20, 0.16], [0.16, 0.18]]
solution = integer_least_squares(float_ambiguities, covariance)
solution
```

## Interpret carefully
- The best integer candidate minimizes the covariance-weighted squared distance.
- The second-best candidate helps describe separation, but the returned ratio is not a universal acceptance threshold.
- A real GNSS workflow must estimate float ambiguities and covariance from carrier-phase observations and validate fixes for integrity.

```python
# Compare with naive component-wise rounding. They need not agree for correlated covariance.
rounded = [int(x + 0.5) if x >= 0 else int(x - 0.5) for x in float_ambiguities]
print('Naive rounded:', rounded)
print('ILS best:', solution['best'])
print('ILS second:', solution['second_best'])
```
