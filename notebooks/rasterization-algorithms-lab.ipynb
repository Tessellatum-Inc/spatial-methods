Compare thin-line, antialiased, conservative traversal and polygon fill contracts. Coordinates here are grid coordinates, not map coordinates.

```python
import sys
sys.path.insert(0, "implementations/python/rasterization")
from rasterize import dda_line, bresenham_line, wu_line, supercover_line, scanline_fill

print('DDA:', dda_line((0, 0), (8, 3)))
print('Bresenham:', bresenham_line((0, 0), (8, 3)))
print('Supercover:', supercover_line((0.5, 0.5), (4.5, 4.5)))
print('Wu intensity samples:', wu_line((0, 0), (8, 3))[:8])
```

```python
import matplotlib.pyplot as plt
grid = scanline_fill([(1, 1), (7, 1), (6, 6), (2, 5)], 8, 9)
plt.imshow(grid, origin='upper', cmap='Greys', interpolation='nearest')
plt.title('Even–odd scanline fill at cell centres'); plt.grid(True); plt.show()
```
